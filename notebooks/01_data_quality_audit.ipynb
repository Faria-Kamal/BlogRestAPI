```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully")
```

```python
import os

files = os.listdir("/content")

print(files)
```

```python
file_path = "/content/online_retail_II.xlsx"
```

```python
excel_file = pd.ExcelFile(file_path)

print("Available sheets:")
print(excel_file.sheet_names)
```

```python
df1 = pd.read_excel(file_path, sheet_name = 0)
df2 = pd.read_excel(file_path, sheet_name = 1)

print("First sheet shape:", df1.shape)
print("Second sheet shape:", df2.shape)
```

```python
df = pd.concat([df1, df2], ignore_index=True)

print("Combined dataset shape:", df.shape)
```

```python
print("First five rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing percentage:")
print(missing_percentage.round(2))
```

```python
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)

negative_quantity = df[df["Quantity"] < 0]

print("Negative quantity records:", len(negative_quantity))

display(negative_quantity.head(10))

```

```python
cancelled_invoices = df[
    df["Invoice"].astype(str).str.startswith("C")
]

print("Invoices beginning with C:", len(cancelled_invoices))

display(cancelled_invoices.head(10))
```

```python
print("Zero or negative prices:")
print((df["Price"] <= 0).sum())

print("\nNegative prices:")
print((df["Price"] < 0).sum())

print("\nZero prices:")
print((df["Price"] == 0).sum())
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

print("Earliest transaction:")
print(df["InvoiceDate"].min())

print("\nLatest transaction:")
print(df["InvoiceDate"].max())
```
